## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
data['listed_in'] = data['listed_in'].str.split(', ')
data = data.explode('listed_in')
data['listed_in'] = data['listed_in'].str.strip()
data['cast'] = data['cast'].str.split(', ')
data = data.explode('cast')
data['cast'] = data['cast'].str.strip()
data['director'] = data['director'].str.split(', ')
data = data.explode('director')
data['director'] = data['director'].str.strip()
data['country'] = data['country'].str.split(', ')
data = data.explode('country')
data['country'] = data['country'].str.strip()
````

````python
data['date_added'] = pd.to_datetime(data['date_added'].str.strip(), errors='coerce').dt.date
print("Date format changed successfully!")
````

````python
data['year_added'] = pd.to_datetime(data['date_added'], errors='coerce').dt.year
print("New column 'year_added' created successfully")
````

````python
data.fillna({"director": "Unknown","cast":"Unknown","country":"Not provided","date_added":"Not provided","rating":"Not provided","year_added":"Not provided"}, inplace=True)
print("All null/missing values filled successfully")
print(f"Remaining null values in dataset:- {data.isnull().sum().sum()}")
````

````python
data['country_category'] = np.where(data['country'] == 'India','India','Other')
print("New column 'country_category' created successfully\n")
print(data['country_category'].value_counts())
````

````python
data = data.reset_index(drop=True)
print("Index reset successfully")
````

#### Chart - 1,2

````python
# Chart - 1,2 visualization code
df_unique = data.drop_duplicates(subset="show_id", keep="first")
type_counts = df_unique["type"].value_counts()

df_india = data[data['country_category'] == 'India']
india = df_india.drop_duplicates(subset="show_id", keep="first")["type"]
india_count=india.value_counts()

fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type': 'domain'}, {'type': 'domain'}]],
   subplot_titles=["<b>Global</b>", "<b>India</b>"]
    )

fig.add_trace(
    go.Pie(
        labels=type_counts.index,
        values=type_counts.values,
        hole=0.2,
        marker=dict(colors=px.colors.qualitative.Pastel)
    ),
    row=1, col=1
)

fig.add_trace(
    go.Pie(
        labels=india_count.index,
        values=india_count.values,
        hole=0.2,
        marker=dict(colors=px.colors.qualitative.Pastel)

    ),
    row=1, col=2
)

fig.update_traces(
    textinfo="percent+label",
    texttemplate="%{label}<br>Count: %{value}<br>%{percent:.0%}",
    textposition="inside",
    insidetextorientation="horizontal"
)

fig.update_layout(
    title="<b>Distribution Comparison of Types (Movies, Tv shows)</b>",
    title_x=0.5,
    template="plotly_white",
    height=550
)

fig.show()
````

````python
# Chart - 2 visualization code
top_5_global = data['listed_in'].value_counts().head(7).index
df_global_filtered = data[data['listed_in'].isin(top_5_global)]
df_global_counts = df_global_filtered.groupby(['listed_in', 'type']).size().reset_index(name='Count')
df_global_counts['Location'] = 'Global'

df_india = data[data['country'] == 'India']
top_5_india = df_india['listed_in'].value_counts().head(7).index
df_india_filtered = df_india[df_india['listed_in'].isin(top_5_india)]
df_india_counts = df_india_filtered.groupby(['listed_in', 'type']).size().reset_index(name='Count')
df_india_counts['Location'] = 'India'

df_combined = pd.concat([df_global_counts, df_india_counts])

fig = px.bar(
    df_combined,
    x='listed_in',
    y='Count',
    color='type',
    barmode='stack',
    text='Count',
    facet_col='Location',
      color_discrete_map={
        'Movie': 'indianred',
        'TV Show': 'saddlebrown'
    }

)

fig.update_xaxes(matches=None, showticklabels=True, categoryorder='total descending')

fig.update_layout(
    title="<b>Categorization of movies and tv shows</b>",
    title_x=0.5,
    template="plotly_white",
    height=550
)
fig.update_yaxes(title_text="<b>Total Number of Titles</b>")
fig.update_xaxes(title_text="<b>Genre / Category</b>", categoryorder='total descending')


fig.show()
````

````python
# Chart - 15 visualization code
df_cleaned = data.drop_duplicates(subset="show_id", keep="first").copy()
df_tv = df_cleaned[df_cleaned['type'] == 'TV Show'].copy()
df_tv['seasons_count'] = df_tv['duration'].str.extract(r'(\d+)').astype(int)
def categorize_seasons(s):
    if s <= 3:
        return '0 to 3 Seasons'
    elif s <= 6:
        return '3 to 6 Seasons'
    else:
        return 'More than 6 Seasons'
df_tv['season_category'] = df_tv['seasons_count'].apply(categorize_seasons)
tv_counts = df_tv['season_category'].value_counts().reset_index()
tv_counts.columns = ['season_category', 'Count']

fig = px.bar(
    tv_counts,
    x='season_category',
    y='Count',
    color='season_category',
    text='Count',
    color_discrete_map={
        '0 to 3 Seasons': 'firebrick',
        '3 to 6 Seasons': 'darkcyan',
        'More than 6 Seasons': 'mediumvioletred'
    }
)

fig.update_layout(
    title="<b>Tv Shows Duration Comparison</b>",
    title_x=0.5,
    template="plotly_white",
    height=500
)
fig.update_yaxes(title_text="<b>Total Number of Titles</b>")
fig.update_xaxes(title_text="<b>Seasons</b>", categoryorder='total descending')

fig.update_traces(textposition='outside')
fig.show()

````

````python
# Chart - 16 visualization code
unique_data = data.drop_duplicates(subset="show_id", keep="first")
df_clean = unique_data[(unique_data['country'] != 'Not provided') & (unique_data['listed_in'] != 'Not provided')]
top_10_countries = df_clean['country'].value_counts().head(10).index
top_10_genres = df_clean['listed_in'].value_counts().head(10).index
df_filtered = df_clean[df_clean['country'].isin(top_10_countries) & df_clean['listed_in'].isin(top_10_genres)]
heatmap_matrix = pd.crosstab(df_filtered['listed_in'], df_filtered['country'])
heatmap_matrix = heatmap_matrix.loc[top_10_genres, top_10_countries]

fig = px.imshow(
    heatmap_matrix,
    text_auto=True,
    labels=dict(x="<b>Country</b>", y="<b>Genre</b>", color="<b>Titles Count</b>"),
    x=heatmap_matrix.columns,
    y=heatmap_matrix.index,
    color_continuous_scale='Reds',
    title="<b>Heatmap: Top 10 Countries vs Top 10 Genres</b>"
)

fig.update_layout(
    title_x=0.5,
    title_y=0.95,
    template="plotly_white",
    height=600,

)

fig.show()
````
